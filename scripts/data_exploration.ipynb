```python
from pathlib import Path
import pandas as pd

# Run this script while within this project at the root folder
ROOT_PATH = Path.cwd().resolve().parent
FOLDER = 'data'
SUB_FOLDER = 'raw'
YEAR = '2025'
FILE_NAME = 'yellow_tripdata_2025-01.parquet'
raw_data = pd.read_parquet(ROOT_PATH / FOLDER / SUB_FOLDER / YEAR /  FILE_NAME)

# Display options
pd.options.display.max_columns = None
pd.options.display.float_format = '{:.4f}'.format
```

## Clean Data

```python
# Apply filter to remove dirty data
cleaned_data = raw_data[
    (raw_data["trip_distance"] > 0.0)
    & (raw_data["fare_amount"] > 0.0)
    & (raw_data["passenger_count"] > 0.0)
    & (raw_data["trip_distance"].notna())
    & (raw_data["fare_amount"].notna())
    & (raw_data["passenger_count"].notna())
    & (raw_data["tpep_dropoff_datetime"] > raw_data["tpep_pickup_datetime"])
]
display(cleaned_data.head(5))
print(cleaned_data.info())
```

## Validate Cleaned Data

```python
# Possible inconsistency in how total_amount column is derived
validation_data = cleaned_data.copy()
validation_data["computed_total_amount"] = (
    cleaned_data["fare_amount"] + cleaned_data["extra"] + cleaned_data["mta_tax"]
    + cleaned_data["tip_amount"] + cleaned_data["tolls_amount"] + cleaned_data["improvement_surcharge"]
    + cleaned_data["congestion_surcharge"] + cleaned_data["Airport_fee"] +  cleaned_data["cbd_congestion_fee"]
)
validation_data["difference"] = validation_data["computed_total_amount"] - validation_data["total_amount"]
validation_data[validation_data["computed_total_amount"] != validation_data["total_amount"]].head(5)
```

## Create dataset with only computed total amount, replace datetime columns with date and hour columns

```python
computed_data = cleaned_data.copy()
computed_data["computed_total_amount"] = (
    cleaned_data["fare_amount"] + cleaned_data["extra"] + cleaned_data["mta_tax"]
    + cleaned_data["tip_amount"] + cleaned_data["tolls_amount"] + cleaned_data["improvement_surcharge"]
    + cleaned_data["congestion_surcharge"] + cleaned_data["Airport_fee"] +  cleaned_data["cbd_congestion_fee"]
)
computed_data["pickup_date"] = computed_data["tpep_pickup_datetime"].dt.date
computed_data["pickup_hour"] = computed_data["tpep_pickup_datetime"].dt.hour
computed_data["dropoff_date"] = computed_data["tpep_dropoff_datetime"].dt.date
computed_data["dropoff_hour"] = computed_data["tpep_dropoff_datetime"].dt.hour
computed_data = computed_data.drop(columns = (['tpep_pickup_datetime','tpep_pickup_datetime','tpep_dropoff_datetime','tpep_dropoff_datetime',
    'fare_amount','extra','mta_tax','tip_amount','tolls_amount','improvement_surcharge','total_amount','congestion_surcharge','Airport_fee','cbd_congestion_fee']))
# computed_data.columns
column_order = [
    'VendorID',
    'PULocationID',
    'pickup_date',
    'pickup_hour',
    'DOLocationID',
    'dropoff_date',
    'dropoff_hour',
    'passenger_count',
    'trip_distance',
    'RatecodeID',
    'store_and_fwd_flag',
    'payment_type',
    'computed_total_amount'
    ]
computed_data = computed_data[column_order]
computed_data
```
